```python
import pandas as pd

file_path = "../data/raw/Dataset Capstone.xlsx"

excel = pd.ExcelFile(file_path)

print("Daftar sheet:")
print(excel.sheet_names)
```

```python
df_menu = pd.read_excel(
    file_path,
    sheet_name="menu"
)

print("Jumlah baris dan kolom:", df_menu.shape)

display(df_menu.head())
```

```python
print("Informasi struktur dataset menu:")
df_menu.info()
```

```python
missing_report = pd.DataFrame({
    "jumlah_kosong": df_menu.isna().sum(),
    "persentase_kosong": (
        df_menu.isna().mean() * 100
    ).round(2)
})

display(missing_report)
```

```python
print("Duplikasi seluruh baris:")
print(df_menu.duplicated().sum())

if "menu_id" in df_menu.columns:
    print("\nDuplikasi menu_id:")
    print(df_menu["menu_id"].dropna().duplicated().sum())
```

```python
print("=== PEMERIKSAAN HARGA MENU ===")

harga = pd.to_numeric(df_menu["price"], errors="coerce")

print("Jumlah harga terisi:", harga.notna().sum())
print("Jumlah harga kosong/tidak numerik:", harga.isna().sum())
print("Jumlah harga negatif:", (harga < 0).sum())
print("Jumlah harga nol:", (harga == 0).sum())

print("\nStatistik harga:")
display(harga.describe())
```

```python
print("=== PEMERIKSAAN KATEGORI MENU ===")

print("Jumlah kategori unik:")
print(df_menu["category"].nunique(dropna=True))

print("\nDaftar kategori:")
display(
    df_menu["category"].value_counts(dropna=False)
)
```

```python
print("=== PEMERIKSAAN RELASI CABANG ===")

df_branches = pd.read_excel(
    file_path,
    sheet_name="branches"
)

valid_branch_ids = set(
    df_branches["branch_id"].dropna()
)

invalid_branches = df_menu[
    ~df_menu["branch_id"].isin(valid_branch_ids)
]

print("Jumlah menu dengan branch_id tidak valid:")
print(len(invalid_branches))

display(
    invalid_branches[
        ["menu_id", "branch_id", "menu_name"]
    ].head(10)
)
```

```python
print("=== 10 MENU DENGAN HARGA TERTINGGI ===")

display(
    df_menu[
        ["menu_id", "branch_id", "menu_name", "category", "price"]
    ]
    .sort_values("price", ascending=False)
    .head(10)
)
```

```python
print("=== DISTRIBUSI MENU PER CABANG ===")

display(
    df_menu.groupby("branch_id")
    .agg(
        jumlah_menu=("menu_id", "count"),
        jumlah_kategori=("category", "nunique"),
        harga_rata_rata=("price", "mean")
    )
    .round(2)
)
```

```python
print("=== DUPLIKASI NAMA MENU PER CABANG ===")

menu_normalized = (
    df_menu["menu_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

duplicate_mask = pd.DataFrame({
    "branch_id": df_menu["branch_id"],
    "menu_name_normalized": menu_normalized
}).duplicated(keep=False)

duplicate_names = df_menu.loc[
    duplicate_mask,
    ["menu_id", "branch_id", "menu_name", "category", "price"]
]

print("Jumlah record yang perlu diperiksa:", len(duplicate_names))
display(duplicate_names)
```

```python
print("=== PERBANDINGAN KATALOG BR001 DAN BR003 ===")

columns = [
    "menu_name",
    "category",
    "price"
]

br001 = (
    df_menu[df_menu["branch_id"] == "BR001"][columns]
    .sort_values(columns)
    .reset_index(drop=True)
)

br003 = (
    df_menu[df_menu["branch_id"] == "BR003"][columns]
    .sort_values(columns)
    .reset_index(drop=True)
)

print("Jumlah menu BR001:", len(br001))
print("Jumlah menu BR003:", len(br003))

print(
    "Apakah katalog identik?",
    br001.equals(br003)
)

if not br001.equals(br003):
    comparison = br001.merge(
        br003,
        on=columns,
        how="outer",
        indicator=True
    )
    display(comparison[comparison["_merge"] != "both"])
```

```python
print("=== DUPLIKASI NAMA DAN KATEGORI ===")

duplicate_category = df_menu.duplicated(
    subset=["branch_id", "menu_name", "category"],
    keep=False
)

result = df_menu.loc[
    duplicate_category,
    [
        "menu_id",
        "branch_id",
        "menu_name",
        "category",
        "price"
    ]
]

print("Jumlah record duplikat:", len(result))

display(result)
```
