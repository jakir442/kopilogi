```python
import pandas as pd
from pathlib import Path

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

file_path = (
    project_root / "data" / "external"
    / "Coffee Shop Sales.xlsx"
)

df_sales = pd.read_excel(
    file_path,
    sheet_name="Transactions"
)

print("Jumlah baris dan kolom:", df_sales.shape)
display(df_sales.head())
```

```python
print("=== STRUKTUR DATASET PENJUALAN ===")
df_sales.info()
```

```python
missing_report = pd.DataFrame({
    "jumlah_kosong": df_sales.isna().sum(),
    "persentase_kosong": (
        df_sales.isna().mean() * 100
    ).round(2)
})

display(missing_report)
```

```python
print("Duplikasi seluruh baris:",
      df_sales.duplicated().sum())

print("Jumlah ID transaksi unik:",
      df_sales["transaction_id"].nunique())

print("Jumlah record:",
      len(df_sales))
```

```python
print("=== PEMERIKSAAN HARGA DAN KUANTITAS ===")

print("Harga kosong:", df_sales["unit_price"].isna().sum())
print("Harga nol:", (df_sales["unit_price"] == 0).sum())
print("Harga negatif:", (df_sales["unit_price"] < 0).sum())

print("\nKuantitas nol:", (df_sales["transaction_qty"] == 0).sum())
print("Kuantitas negatif:", (df_sales["transaction_qty"] < 0).sum())

print("\nStatistik harga:")
display(df_sales["unit_price"].describe())

print("\nStatistik kuantitas:")
display(df_sales["transaction_qty"].describe())
```

```python
print("=== PEMERIKSAAN TANGGAL DAN WAKTU ===")

print("Tanggal paling awal:")
print(df_sales["transaction_date"].min())

print("\nTanggal paling akhir:")
print(df_sales["transaction_date"].max())

print("\nContoh waktu transaksi:")
display(df_sales["transaction_time"].head(10))

print("\nTipe data waktu:")
display(
    df_sales["transaction_time"]
    .map(type)
    .value_counts()
)
```

```python
print("=== PEMERIKSAAN KATEGORI PRODUK ===")

print("Jumlah kategori unik:")
print(df_sales["product_category"].nunique())

print("\nDistribusi kategori:")
display(
    df_sales["product_category"]
    .value_counts(dropna=False)
)

print("\nJumlah product_type unik:")
print(df_sales["product_type"].nunique())

print("\nJumlah product_detail unik:")
print(df_sales["product_detail"].nunique())
```

```python
print("=== PEMERIKSAAN LOKASI TOKO ===")

store_summary = (
    df_sales.groupby(
        ["store_id", "store_location"]
    )
    .agg(
        jumlah_record=("transaction_id", "count"),
        jumlah_produk=("product_id", "nunique")
    )
    .reset_index()
)

display(store_summary)

print("\nJumlah lokasi unik:")
print(df_sales["store_location"].nunique())
```

```python
print("=== PEMERIKSAAN KONSISTENSI PRODUK ===")

product_check = (
    df_sales.groupby("product_id")
    .agg(
        jumlah_kategori=("product_category", "nunique"),
        jumlah_jenis=("product_type", "nunique"),
        jumlah_detail=("product_detail", "nunique"),
        jumlah_harga=("unit_price", "nunique")
    )
)

print("Product ID dengan lebih dari satu kategori:")
print((product_check["jumlah_kategori"] > 1).sum())

print("\nProduct ID dengan lebih dari satu jenis:")
print((product_check["jumlah_jenis"] > 1).sum())

print("\nProduct ID dengan lebih dari satu detail:")
print((product_check["jumlah_detail"] > 1).sum())

print("\nProduct ID dengan lebih dari satu harga:")
print((product_check["jumlah_harga"] > 1).sum())

display(product_check.head(10))
```

```python
print("=== PEMERIKSAAN VARIASI HARGA PRODUK ===")

# Identifikasi produk yang memiliki lebih dari satu harga
variable_price_ids = product_check.index[
    product_check["jumlah_harga"] > 1
]

price_variations = (
    df_sales[
        df_sales["product_id"].isin(variable_price_ids)
    ]
    .groupby([
        "product_id",
        "product_category",
        "product_type",
        "product_detail"
    ])
    .agg(
        jumlah_harga=("unit_price", "nunique"),
        harga_minimum=("unit_price", "min"),
        harga_maksimum=("unit_price", "max"),
        jumlah_record=("transaction_id", "count")
    )
    .reset_index()
    .sort_values("jumlah_harga", ascending=False)
)

display(price_variations)
```

```python
print("=== VARIASI HARGA BERDASARKAN LOKASI ===")

price_by_store = (
    df_sales[
        df_sales["product_id"].isin(variable_price_ids)
    ]
    .groupby([
        "product_id",
        "store_location"
    ])
    .agg(
        jumlah_harga=("unit_price", "nunique"),
        harga_minimum=("unit_price", "min"),
        harga_maksimum=("unit_price", "max")
    )
    .reset_index()
)

display(price_by_store.head(30))
```

```python
print("=== 10 PRODUK DENGAN HARGA TERTINGGI ===")

highest_prices = (
    df_sales[
        [
            "product_id",
            "product_category",
            "product_type",
            "product_detail",
            "unit_price"
        ]
    ]
    .drop_duplicates()
    .sort_values("unit_price", ascending=False)
    .head(10)
)

display(highest_prices)
```
