```python
import pandas as pd
from pathlib import Path

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

file_path = (
    project_root
    / "data"
    / "raw"
    / "Dataset Capstone.xlsx"
)

df_info = pd.read_excel(
    file_path,
    sheet_name="general_information"
)

print("=== DATASET GENERAL INFORMATION ===")
print("Jumlah baris dan kolom:", df_info.shape)

display(df_info.head())
```

```python
print("=== STRUKTUR DATASET ===")

df_info.info()

print("\nDaftar kolom:")
print(df_info.columns.tolist())
```

```python
print("=== PEMERIKSAAN MISSING VALUES ===")

missing_report = pd.DataFrame({
    "jumlah_kosong": df_info.isna().sum(),
    "persentase_kosong": (
        df_info.isna().mean() * 100
    ).round(2)
})

display(missing_report)
```

```python
print("=== PEMERIKSAAN DUPLIKASI ===")

print(
    "Duplikasi seluruh baris:",
    df_info.duplicated().sum()
)

print(
    "Duplikasi info_id:",
    df_info["info_id"].duplicated().sum()
)

print(
    "Duplikasi info_key:",
    df_info["info_key"].duplicated().sum()
)
```

```python
print("=== STATUS VERIFIKASI ===")

display(
    df_info["verification_status"]
    .value_counts(dropna=False)
)

print("\n=== JENIS DATA ===")

display(
    df_info["data_type"]
    .value_counts(dropna=False)
)
```

```python
print("=== PEMERIKSAAN SUMBER INFORMASI ===")

display(
    df_info[
        [
            "info_id",
            "info_key",
            "source",
            "source_url",
            "accessed_at"
        ]
    ]
)

print("\nDistribusi sumber:")
display(
    df_info["source"].value_counts(dropna=False)
)
```

```python
print("=== PEMERIKSAAN KONTEN ===")

content_check = df_info[
    ["info_id", "info_key", "title", "content"]
].copy()

content_check["content_length"] = (
    content_check["content"]
    .astype("string")
    .str.strip()
    .str.len()
)

content_check["is_content_empty"] = (
    content_check["content_length"].fillna(0) == 0
)

display(
    content_check[
        [
            "info_id",
            "info_key",
            "content_length",
            "is_content_empty"
        ]
    ]
)

print(
    "Jumlah konten kosong:",
    content_check["is_content_empty"].sum()
)
```

```python
print("=== VALIDASI AKHIR GENERAL INFORMATION ===")

# Pemeriksaan duplikasi
duplicate_rows = df_info.duplicated().sum()
duplicate_ids = df_info["info_id"].duplicated().sum()
duplicate_keys = df_info["info_key"].duplicated().sum()

# Pemeriksaan konten kosong atau hanya spasi
empty_content = (
    df_info["content"]
    .astype("string")
    .str.strip()
    .fillna("")
    .eq("")
    .sum()
)

print("Jumlah record:", len(df_info))
print("Duplikasi seluruh baris:", duplicate_rows)
print("Duplikasi info_id:", duplicate_ids)
print("Duplikasi info_key:", duplicate_keys)
print("Konten kosong:", empty_content)

print("\nPanjang konten:")
display(
    df_info.assign(
        content_length=(
            df_info["content"]
            .astype("string")
            .str.strip()
            .str.len()
        )
    )[["info_id", "info_key", "content_length"]]
)
```

```python
from pathlib import Path
import pandas as pd

reports_dir = project_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

quality_report = pd.DataFrame([
    {
        "check_id": "GI001",
        "aspect": "Missing Values",
        "finding": "Tidak ditemukan nilai kosong",
        "affected_records": int(df_info.isna().any(axis=1).sum()),
        "decision": "Tidak diperlukan imputasi"
    },
    {
        "check_id": "GI002",
        "aspect": "Duplicate ID",
        "finding": f"{duplicate_ids} info_id duplikat",
        "affected_records": int(duplicate_ids),
        "decision": "Tinjau jika ditemukan duplikasi"
    },
    {
        "check_id": "GI003",
        "aspect": "Duplicate Topic",
        "finding": f"{duplicate_keys} info_key duplikat",
        "affected_records": int(duplicate_keys),
        "decision": "Periksa apakah topik berulang"
    },
    {
        "check_id": "GI004",
        "aspect": "Content Completeness",
        "finding": f"{empty_content} konten kosong",
        "affected_records": int(empty_content),
        "decision": "Tidak mengarang konten pengganti"
    },
    {
        "check_id": "GI005",
        "aspect": "PIC Verification",
        "finding": "Informasi masih menunggu verifikasi PIC",
        "affected_records": int(
            df_info["verification_status"].eq("pending_pic").sum()
        ),
        "decision": "Pertahankan status pending_pic"
    },
    {
        "check_id": "GI006",
        "aspect": "Source Traceability",
        "finding": "Referensi wawancara tercatat",
        "affected_records": len(df_info),
        "decision": "Periksa kesesuaian konten dengan bukti wawancara"
    }
])

output_path = reports_dir / "general_information_quality_report.csv"

quality_report.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

display(quality_report)

print("Laporan berhasil disimpan:", output_path)
```
